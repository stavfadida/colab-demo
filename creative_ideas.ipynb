# Creative uses for everyday objects

Enter an everyday object and get **three unusual, practical ideas** for it. The notebook then lets you try another object or exit.

Run the cells from top to bottom. You will enter your API key privately when prompted.

```python
# Install the OpenAI Python SDK in this Colab session.
%pip -q install openai
```

```python
from getpass import getpass
from openai import OpenAI

# getpass hides the key while you type it. The key is not saved in this notebook.
api_key = getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)
```

```python
# Keep suggesting ideas until the user chooses to exit.
while True:
    item = input("Name an everyday object (or type 'exit' to quit): " ).strip()

    # Stop when the user types exit.
    if item.lower() == "exit":
        print("Goodbye!")
        break

    # Ask again if the input is blank.
    if not item:
        print("Please enter an object name.\n")
        continue

    # Ask the model for three distinct ideas.
    response = client.responses.create(
        model="gpt-6-luna",
        input=(
            f"Suggest exactly THREE unusual but practical uses for this everyday object: {item}. "
            "Number each idea and explain each in one or two clear sentences. Make them safe and realistic."
        ),
    )

    print("\nHere are three ideas:\n")
    print(response.output_text)
    print()
```
