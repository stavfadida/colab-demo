# One unusual use for an everyday object

Tell the notebook an everyday object. The OpenAI Responses API will suggest **one unusual, practical use** for it.

Run the cells from top to bottom. You will enter your API key privately when prompted.

```python
# Install the OpenAI Python SDK in this Colab session.
%pip -q install openai
```

```python
from getpass import getpass
from openai import OpenAI

# getpass hides the key while you type it. The key is not saved in this notebook.
api_key = getpass("Enter your OpenAI API key: ")
client = OpenAI(api_key=api_key)
```

```python
# Type an everyday object, such as a mug, shoelace, or paper clip.
item = input("Name an everyday object: " ).strip()

if not item:
    raise ValueError("Please enter an everyday object and run this cell again.")
```

```python
# Ask for exactly one creative idea, with a short explanation.
response = client.responses.create(
    model="gpt-6-luna",
    input=(
        f"Suggest ONE unusual but practical use for this everyday object: {item}. "
        "Give the idea in one or two clear sentences. Make it safe and realistic."
    ),
)

print(response.output_text)
```
